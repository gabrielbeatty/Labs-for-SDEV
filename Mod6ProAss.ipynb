13.1 Write the current date as a string to the text file today.txt.

```python
import datetime

# Get Todays Current Date
today_string = datetime.date.today().strftime("%Y-%m-%d")

# Write that date to today.txt
with open("today.txt", "w") as f:
    f.write(today_string)

print(f"Successfully wrote {today_string} to today.txt")
```

13.3 Parse the date from today_string.

```python
from datetime import datetime

# Open the file and read into a string variable
with open("today.txt", "r") as f:
    today_string = f.read()

# Parse the string into a datetime object**Remember strptime**
# %Y = 4 digit year, %m = 2 digit month, %d = 2 digit day
date_object = datetime.strptime(today_string, "%Y-%m-%d")

# Verify
print(f"Object: {date_object}")
print(f"Type: {type(date_object)}")
```

15.1 Use multiprocessing to create three separate processes. Make each one wait a random number of seconds between zero and one, print the current time, and then exit.

**Writing to a .py file to mitigate issues using Jupyter NB**

```python
%%writefile multiprocessing_worker.py
import time
import random
from datetime import datetime

def worker_task(process_num):
    # Wait a random number of seconds between 0 and 1
    wait_time = random.random()
    time.sleep(wait_time)
    
    # Get current time
    current_time = datetime.now().strftime("%H:%M:%S.%f")[:-3]
    # The -3 sets the precision for the millisecond. Truncates.
    
    # Print result and exit
    print(f"Process {process_num}: Waited {wait_time:.3f}s | Current Time: {current_time}", flush=True)
    # 3f formats to 3 decimal places. Similar to above truncation.
```

```python
import multiprocessing
from multiprocessing_worker import worker_task

if __name__ == "__main__":
    processes = []
    
    # Create 3 processes and start
    for i in range(3):
        p = multiprocessing.Process(target=worker_task, args=(i+1,))
        processes.append(p)
        p.start()
    
    # Exit when all processes completed
    for p in processes:
        p.join()

    print("All processes are completed.")
```
